```python
df=spark.read.option('header',True).csv('/Volumes/workspace/default/product/coupon_redempt.csv')
df.display()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import DecimalType,IntegerType,StringType

df1=df.withColumn('household_key',F.col('household_key').cast(IntegerType()))\
     .withColumn('DAY',F.col('DAY').cast(IntegerType()))\
     .withColumn('COUPON_UPC',F.col('COUPON_UPC').cast('string'))\
     .withColumn('CAMPAIGN',F.col('CAMPAIGN').cast(IntegerType()))\
     .withColumn('ImportDateTime',F.current_timestamp())

df1.display()
```

```python
df1.write.format('delta').option('mergeSchema', 'true').mode('overwrite').save('/Volumes/workspace/default/product/coupon_redempt.delta')
```

```python
df1.write.format('delta').mode('overwrite').mode('overwrite').saveAsTable('default.coupon_redempt')
```
