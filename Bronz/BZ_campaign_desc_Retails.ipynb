```python
df=spark.read.option('header',True).csv('/Volumes/workspace/default/product/campaign_desc.csv')
df.display()
```

```python

from pyspark.sql import functions as F
from pyspark.sql.types import DecimalType,IntegerType,StringType

df1=df.withColumn('DESCRIPTION',F.col('DESCRIPTION').cast('string'))\
     .withColumn('CAMPAIGN',F.col('CAMPAIGN').cast(IntegerType()))\
     .withColumn('START_DAY',F.col('START_DAY').cast(IntegerType()))\
     .withColumn('END_DAY',F.col('END_DAY').cast(IntegerType()))

df1.show()
```

```python
df1.write.format('delta').mode('overwrite').save('/Volumes/workspace/default/product/campaign_desc')
#df3=df2.drop('CustomerPhone')
#df3.display()
```
