```python
df=spark.read.option('header',True).csv('/Volumes/workspace/default/product/transaction_data_12345.csv')
df.display()
```

```python

from pyspark.sql import functions as F
from pyspark.sql.types import DecimalType,IntegerType,StringType

df1=df.withColumn('household_key',F.col('household_key').cast(IntegerType()))\
     .withColumn('BASKET_ID',F.col('BASKET_ID').cast('string'))\
     .withColumn('DAY',F.col('DAY').cast(IntegerType()))\
     .withColumn('PRODUCT_ID',F.col('PRODUCT_ID').cast(IntegerType()))\
     .withColumn('QUANTITY',F.col('QUANTITY').cast(IntegerType()))\
     .withColumn('STORE_ID',F.col('STORE_ID').cast(IntegerType()))\
     .withColumn('RETAIL_DISC',F.col('RETAIL_DISC').cast(DecimalType(10,3)))\
     .withColumn('TRANS_TIME',F.col('TRANS_TIME').cast(IntegerType()))\
     .withColumn('WEEK_NO',F.col('WEEK_NO').cast(IntegerType()))\
     .withColumn('COUPON_DISC',F.col('COUPON_DISC').cast('string'))\
     .withColumn('COUPON_MATCH_DISC',F.col('COUPON_MATCH_DISC').cast('string'))\
     .withColumn('ImportDateTime',F.current_timestamp())

df1.display()
```

```python
df1.write.format('delta').option('mergeSchema', 'true').mode('overwrite').save('/Volumes/workspace/default/product/transaction_data.delta')
```

```python
df1.write.format('delta').mode('overwrite').mode('overwrite').saveAsTable('default.transaction_data')
```
