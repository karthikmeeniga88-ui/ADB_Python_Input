```python
df=spark.read.option('header',True).csv('/Volumes/workspace/default/product/hh_demographic.csv')
df.display()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import DecimalType,IntegerType,StringType

df1=df.withColumn('AGE_DESC',F.col('AGE_DESC').cast('string'))\
     .withColumn('MARITAL_STATUS_CODE',F.col('MARITAL_STATUS_CODE').cast('string'))\
     .withColumn('INCOME_DESC',F.col('INCOME_DESC').cast('string'))\
     .withColumn('HOMEOWNER_DESC',F.col('HOMEOWNER_DESC').cast('string'))\
     .withColumn('HH_COMP_DESC',F.col('HH_COMP_DESC').cast('string'))\
     .withColumn('HOUSEHOLD_SIZE_DESC',F.col('HOUSEHOLD_SIZE_DESC').cast('string'))\
     .withColumn('KID_CATEGORY_DESC',F.col('KID_CATEGORY_DESC').cast('string'))\
     .withColumn('household_key',F.col('household_key').cast(IntegerType()))\
     .withColumn('ImportDateTime',F.current_timestamp())

df1.display()
```

```python
df1.write.format('delta').option('mergeSchema', 'true').mode('overwrite').save('/Volumes/workspace/default/product/Demographic.delta')
```
