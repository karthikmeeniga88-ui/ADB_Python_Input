```python
df=spark.read.option('header',True).csv('/Volumes/workspace/default/input/Customer.csv')
df.show()
```

```python

from pyspark.sql import functions as F
from pyspark.sql.types import DecimalType,IntegerType,StringType

df1=df.withColumn('COUPON_UPC',F.col('COUPON_UPC').cast('string'))\
     .withColumn('PRODUCT_ID',F.col('PRODUCT_ID').cast(IntegerType()))\
     .withColumn('CAMPAIGN',F.col('CAMPAIGN').cast(IntegerType()))\
     .withColumn('ImportDateTime',F.current_timestamp())

df1.display()
```

```python
df1.write.format('delta').option('mergeSchema', 'true').mode('overwrite').save('/Volumes/workspace/default/product/coupon.delta')
```

```python
df1.write.format('delta').mode('overwrite').mode('overwrite').saveAsTable('default.coupon')
```
