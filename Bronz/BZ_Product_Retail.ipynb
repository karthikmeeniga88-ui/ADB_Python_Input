```python
df=spark.read.option('header',True).csv('/Volumes/workspace/default/product/product.csv')
df.display()
```

```python

from pyspark.sql import functions as F
from pyspark.sql.types import DecimalType,IntegerType,StringType

df1=df.withColumn('PRODUCT_ID',F.col('PRODUCT_ID').cast(IntegerType()))\
     .withColumn('MANUFACTURER',F.col('MANUFACTURER').cast(IntegerType()))\
     .withColumn('DEPARTMENT',F.col('DEPARTMENT').cast('string'))\
     .withColumn('BRAND',F.col('BRAND').cast('string'))\
     .withColumn('COMMODITY_DESC',F.col('COMMODITY_DESC').cast('string'))\
     .withColumn('SUB_COMMODITY_DESC',F.col('SUB_COMMODITY_DESC').cast('string'))\
     .withColumn('CURR_SIZE_OF_PRODUCT',F.col('CURR_SIZE_OF_PRODUCT').cast('string'))

df1.display()
```

```python
df1.display()
```

```python
df1.write.format('delta').mode('overwrite').save('/Volumes/workspace/default/product/product_camp.delta')
```
