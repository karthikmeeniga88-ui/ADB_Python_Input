```python
df=spark.read.option('header',True).csv('/Volumes/workspace/default/product/causal_data.csv')
df.display()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import DecimalType,IntegerType,StringType

df1=df.withColumn('PRODUCT_ID',F.col('PRODUCT_ID').cast(IntegerType()))\
     .withColumn('STORE_ID',F.col('STORE_ID').cast(IntegerType()))\
     .withColumn('WEEK_NO',F.col('WEEK_NO').cast(IntegerType()))\
     .withColumn('display',F.col('display').cast('string'))\
     .withColumn('mailer',F.col('mailer').cast('string'))\
     .withColumn('ImportDateTime',F.current_timestamp())

df1.display()
```

```python
df1.write.format('delta').option('mergeSchema', 'true').mode('overwrite').save('/Volumes/workspace/default/product/causal_data.delta')
```
