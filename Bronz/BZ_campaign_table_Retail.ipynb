```python
df=spark.read.option('header',True).csv('/Volumes/workspace/default/product/campaign_table.csv')
df.display()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import DecimalType,IntegerType,StringType

df1=df.withColumn('DESCRIPTION',F.col('DESCRIPTION').cast('string'))\
     .withColumn('household_key',F.col('household_key').cast(IntegerType()))\
     .withColumn('CAMPAIGN',F.col('CAMPAIGN').cast(IntegerType()))\
     .withColumn('ImportDateTime',F.current_timestamp())

df1.display()
```

```python
df1.write.format('delta').option('mergeSchema', 'true').mode('overwrite').save('/Volumes/workspace/default/product/campaign_tabledelta')
```

```python
df1.write.format('delta').mode('overwrite').mode('overwrite').saveAsTable('default.campaign_table')
```
